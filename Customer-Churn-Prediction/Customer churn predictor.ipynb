# We’ll: compare multiple ML models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
models = {
    'Logistic Regression': LogisticRegression(),
    'KNN': KNeighborsClassifier(),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'XGBoost': XGBClassifier()
}
```

```python
results=[]
```

```python


from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score

for name ,model in models.items():
  model.fit(X_train_scaled,y_train)

  y_pred = model.predict(X_test_scaled)

  accuracy = accuracy_score(y_test,y_pred)

  f1 = f1_score(y_test , y_pred)

  auc = roc_auc_score(y_test , y_pred)

  results.append({
      'Model':name,
      'Acuuracy':accuracy,
      'F1 Score':f1,
      'ROC-AUC':auc
  })
```

```python
result_df = pd.DataFrame(results)

result_df
```

**sorting best model**

```python
result_df.sort_values(
    by = 'ROC-AUC',
    ascending = False,
)
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x = 'Model',
    y = 'ROC-AUC',
    data = result_df
)

plt.title('Model Comparison')
plt.show()
```

#Hyperparameter tuning for Logistic Regression

```python
from sklearn.model_selection import GridSearchCV

params_grid = {
  'C' : [0.01 , 0.1 , 1 , 10 , 100],
  'penalty' : ['l1' , 'l2'],
  'solver' : ['liblinear']
}

grid = GridSearchCV(
    LogisticRegression(),
    params_grid,
    cv=5,
    scoring  = 'roc_auc'
)

grid.fit(X_train_scaled ,y_train)

```

```python
print(grid.best_params_)
```

```python
print(grid.best_score_)
```

**predict**

```python
best_lr  = grid.best_estimator_

# prediction probabilities
y_prob = best_lr.predict_proba(X_test_scaled)[:,1]

# final predictions
y_pred = best_lr.predict(X_test_scaled)


auc = roc_auc_score(y_test , y_prob)

print("AUC = " , auc)
```

**Accuracy**

Threshold Optimization

Then we'll likely push Recall:

52%
↓
65–75%

while keeping decent performance.

 churn prediction cares more about finding churners than maximizing raw accuracy.

 By default,
  the computer uses 50% as the line:
  Above 50%? It labels it a Yes.Below 50%? It labels it a No.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.3,0.35,0.4,0.45,0.5,0.55,0.6]

for t in thresholds:

    y_pred_t = (y_prob >= t).astype(int)

    print(f"\nThreshold = {t}")

    print("Precision:", round(precision_score(y_test,y_pred_t),3))
    print("Recall   :", round(recall_score(y_test,y_pred_t),3))
    print("F1 Score :", round(f1_score(y_test,y_pred_t),3))
```

```python
best_threshold = 0.3
final_predictions = (y_prob>= best_threshold).astype(int)
```

```python
print(classification_report(y_test , final_predictions))
```

```python
print(confusion_matrix(y_test, final_predictions))
```
